```python
data=[('Tom',23),('John',30),('Jane',14),('Matt',44)]
columns = ['name','age']
df=spark.createDataFrame(data,columns)
df.show() #show is a action in dataframe whcih executes and gives the code to driver node
#show,count,collect are some of the action items in dataframe which executes the code and gives the output
```

```python
#DAG-Directed Acyclic Graph
df.printSchema() #gives structure of df
```

```python
from pyspark.sql.functions import col, lit
#df.withColumn('gender', lit('Female'))
df=df.withColumn('gender', lit('Male'))
df.show()
```

```python
df=df.withColumn('modified_age',col('age')+10)
df.show()
```

```python
df=df.withColumnRenamed('name','customer_name')
df.show()
```

```python
df=df.drop('age')
df.show()
```

```python
from pyspark.sql.functions import when
```

```python
from pyspark.sql.functions import when
df=df.withColumn('department',when(col('customer_name')=='Tom','IT').otherwise(lit('HR')))
df.show()
```

```python
df=df.orderBy(col('customer_name').desc())
df.show()
```

```python
emp_df = spark.createDataFrame([(1, 'Tom', 23, 10), (2, 'John', 30, 10), (3, 'Jane', 14, 12), (4, 'Matt', 44, 12)], ['empid', 'name', 'age', 'deptid'])
emp_df.show()
```

```python
dept_df = spark.createDataFrame([(10, 'IT'), (12, 'Sales')], ['deptid', 'deptname'])
dept_df.show()
```

```python
emp_df.join(dept_df,on='deptid',how='inner')
emp_dept_df=emp_df.join(dept_df,on='deptid',how='inner')
emp_dept_df.show()
```

```python

emp_dept_df=emp_df.join(dept_df,emp_df.deptid==dept_df.deptid,'inner')
emp_dept_df.show()
```

```python
a=emp_df.alias('e') # e is alias as emp_df
emp_dept_df2=a.join(dept_df,a.deptid==dept_df.deptid)
emp_dept_df2.show()
d=dept_df.alias('d')
e=emp_df.alias('e')
d
```

```python
d = dept_df.alias('d')
emp_dept_df2 = a.join(d, a.deptid == d.deptid, 'inner').select(a.empid, a.name, a.age, d.deptname)
emp_dept_df2.show()
```

```python
empjoined=emp_df.alias('em').join(dept_df.alias('dp'),col('em.deptid')==col('dp.deptid'),how='inner')
empjoined.show()
```

```python
empjoined.groupBy('deptname').count().show()
```

```python
from pyspark.sql.functions import sum,count
empjoined.groupBy('deptname').agg(sum('age').alias('totalage'),count('deptname').alias('deptcount')).show()
```
